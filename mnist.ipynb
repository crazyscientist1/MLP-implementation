# MNIST Implementation

**Loading Data**

```python
from keras.datasets import mnist
from matplotlib import pyplot as plt
import numpy as np
```

```python
(train_X, train_y), (test_X, test_y) = mnist.load_data()
print('train_X: ' + str(train_X.shape))
print('train_y: ' + str(train_y.shape))
print('test_X: '  + str(test_X.shape))
print('test_y: '  + str(test_y.shape))
```

```python
train_y, test_y = np.eye(10)[train_y], np.eye(10)[test_y]
```

```python
print('train_y: '  + str(train_y.shape))
print('test_y: '  + str(test_y.shape))
```

**Showing Example**

```python
plt.imshow(train_X[0])
print(train_y[0])
```

```python
train_X[0].flatten().shape
```

```python
from models import MLP, Layer, Linear, ReLU, MSE, Tanh, Softmax, Cross_Entropy

mlp = MLP(784, Cross_Entropy)
mlp.add(Layer(784, ReLU))
mlp.add(Layer(500, ReLU))
mlp.add(Layer(500, ReLU))
mlp.add(Layer(200, ReLU))
mlp.add(Layer(50, ReLU))
mlp.add(Layer(10, Softmax))
```

```python
train_y.shape
```

```python
mlp.train(train_X.reshape(len(train_X), 784) / 255.0, train_y, 10000, 500, 0.001)
```

```python
plt.imshow(test_X[50])
test_y[50]

np.argmax(test_X)
```

```python
print(f"Accuracy: {np.sum((np.eye(10)[np.argmax(mlp.forward(test_X.reshape(-1,784)/255.0),axis = 1)] * test_y) == 1)/np.shape(test_y)[0]")
```
