```python
import numpy as np
from bokeh.plotting import figure, curdoc
from bokeh.models import ColumnDataSource
from model import *
import elements

# Data
x = np.linspace(-10, 10, 1000)

source = ColumnDataSource(data=dict(x=x, y=np.zeros_like(x)))

p = figure(width=1000, height=500)
p.line(x, np.sin(x), line_width=3)
p.line('x', 'y', source=source, line_width=3, line_color="red")

# Build model
mlp = MLP(1, elements.MSE)
mlp.add(Layer(6, elements.ReLU))
mlp.add(Layer(6, elements.ReLU))
mlp.add(Layer(6, elements.ReLU))
mlp.add(Layer(1, elements.Linear))

def update():
    idx = np.random.randint(0, len(x))
    x_input = x[idx].reshape(1,1)
    y_target = np.sin(x_input)

    mlp.training_step(x_input, y_target, 0.001)

    out = mlp.forward(x.reshape(-1,1))
    source.data = dict(x=x, y=out[:,0])

curdoc().add_root(p)
curdoc().add_periodic_callback(update, 50)
```
